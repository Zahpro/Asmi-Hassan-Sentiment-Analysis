```python
import re

def clean_arabic_expert(text):
    if not isinstance(text, str): return ""

    # Normalize common Arabic variations
    text = re.sub("[إأآ]", "ا", text)
    text = re.sub("ة", "ه", text)
    text = re.sub("رر", "ر", text)
    # Remove elongation (Tashkeel)
    text = re.sub(r'(.)\1+', r'\1', text)

    # Preserve letters and specific emojis, remove punctuation/symbols
    emoji_pattern = "\U0001F600-\U0001F64F\U0001F300-\U0001F5FF"
    clean_pattern = f'[^\\w\\s\\u0600-\\u06FF{emoji_pattern}]'
    text = re.sub(clean_pattern, ' ', text)

    # Remove Arabic diacritics (Tashkeel)
    tashkeel = ["\u064B", "\u064C", "\u064D", "\u064E", "\u064F", "\u0650", "\u0651", "\u0652"]
    for move in tashkeel: text = text.replace(move, "")

    return ' '.join(text.split()).strip()

df_all['text'] = df_all['text'].apply(clean_arabic_expert)
print("Arabic linguistic cleaning completed successfully")
```

```python
# Define sentiment lexicons for Iraqi dialect (Expanded)
pos_words = [
    'فد شي', 'العظيم', 'سنمضي', 'مبدع', 'حلو', 'اجمل', 'عاشت', 'يجنن',
    'افضل', 'راقي', 'استمر', 'تجنن', 'احلى', 'واقعي', 'ممتاز', 'فنان', 'ونعم', 'بطل', 'توب', 'معدل', 'ذهب', 'كفو', 'روعة', 'موفق',
    'يخبل', 'فخورين', 'رهيب'
]
neg_words = [
    'محلو', 'لعبان نفس', 'فاشل', 'خربطتوها', 'ضعيف', 'مضيعة وكت', 'موصلت', 'سيء', 'خساره', 'ملل', 'مهزله', 'تعبان', 'ضياع', 'خربت','هدام',
    'موصحيح', 'كذب', 'تلفيق', 'سخيف', 'ياوسخ', 'ظالمين', 'كذابين', 'مكروه'
]

# Define multi-word phrases for sentiment analysis
pos_phrases = [
    'عاش السانك', 'وشعب سعيد', 'محمد وال محمد', 'يرفع الراس',
    'الله يرحم', 'اله يرحم', 'عاشت ايدكم', 'رحم الله' ]

neg_phrases = [
    'لعنه الله', 'لعنه اله', 'خرب روحك صدام', 'الهم العن صدام',
    'حسبي الله', 'حسبي اله', 'الله ينتقم', 'اله ينتقم',
    'يلعن', 'اسخف شي', 'لا بارك الله'
]

def get_sentiment(text):
    score = 0
    if not isinstance(text, str): return 'Neutral'

    # Convert text to lowercase for case-insensitive matching if applicable
    # (Not strictly necessary for Arabic as it's often case-insensitive)
    # text_lower = text.lower()

    # Check for negative phrases first
    for phrase in neg_phrases:
        if phrase in text:
            score -= 2 # Give higher weight to phrases

    # Check for positive phrases
    for phrase in pos_phrases:
        if phrase in text:
            score += 2 # Give higher weight to phrases

    # Split text into words for single-word lexicon matching
    words = text.split()

    # Check for individual words (only if they are not part of an already matched phrase to avoid double counting)
    for word in words:
        if word in pos_words:
            score += 1
        if word in neg_words:
            score -= 1

    if score > 0:
        return 'Positive'
    elif score < 0:
        return 'Negative'
    else:
        return 'Neutral'

# Apply sentiment analysis
df_all['sentiment'] = df_all['text'].apply(get_sentiment)

# Display updated results
sentiment_counts = df_all['sentiment'].value_counts()
print("\nTotal Sentiment Analysis Results (Updated Lexicon):")
print(sentiment_counts)
```

```python
from sklearn.feature_extraction.text import CountVectorizer
from collections import Counter

def get_saddam_ngrams(corpus, n, top_k=10):
    # Filter comments containing 'صدام'
    saddam_corpus = [text for text in corpus if 'صدام' in text]

    if not saddam_corpus:
        return []

    vec = CountVectorizer(ngram_range=(n, n)).fit(saddam_corpus)
    bag_of_words = vec.transform(saddam_corpus)
    sum_words = bag_of_words.sum(axis=0)
    words_freq = [(word, sum_words[0, idx]) for word, idx in vec.vocabulary_.items()]
    words_freq = sorted(words_freq, key=lambda x: x[1], reverse=True)
    return words_freq[:top_k]

# Filter neutral comments for analysis
neutral_saddam_comments = df_all[(df_all['sentiment'] == 'Neutral') & (df_all['text'].str.contains('صدام'))]['text']

print("⚫ TOP 10 BIGRAMS (2-word phrases) containing 'صدام' in neutral comments:")
common_saddam_bigrams = get_saddam_ngrams(neutral_saddam_comments, n=2, top_k=10)
for phrase, freq in common_saddam_bigrams:
    print(f"{phrase}: {freq}")

print("\n⚫ TOP 10 TRIGRAMS (3-word phrases) containing 'صدام' in neutral comments:")
common_saddam_trigrams = get_saddam_ngrams(neutral_saddam_comments, n=3, top_k=10)
for phrase, freq in common_saddam_trigrams:
    print(f"{phrase}: {freq}")
```

```python
!pip install gensim
from gensim import corpora
from gensim.models import LdaModel

def get_topics_by_sentiment(sentiment_label, num_topics=3):
    # Filter data
    subset = df_all[df_all['sentiment'] == sentiment_label]['text']
    if len(subset) < 10:
        return f"Not enough data for {sentiment_label} sentiment."

    # Tokenize and filter short comments
    tokenized = [text.split() for text in subset if len(text) > 10]
    if not tokenized:
        return f"No valid text found for {sentiment_label} sentiment after filtering."

    # Build LDA
    dictionary = corpora.Dictionary(tokenized)
    corpus = [dictionary.doc2bow(text) for text in tokenized]
    lda = LdaModel(corpus=corpus, id2word=dictionary, num_topics=num_topics, passes=10, random_state=42)

    return lda.print_topics()

print("--- TOPICS IN POSITIVE COMMENTS ---")
pos_topics = get_topics_by_sentiment('Positive')
if isinstance(pos_topics, list):
    for idx, topic in pos_topics:
        print(f"Topic {idx}: {topic}")
else:
    print(pos_topics)

print("\n--- TOPICS IN NEGATIVE COMMENTS ---")
neg_topics = get_topics_by_sentiment('Negative')
if isinstance(neg_topics, list):
    for idx, topic in neg_topics:
        print(f"Topic {idx}: {topic}")
else:
    print(neg_topics)
```

```python
# 1. Define the list of target entities (names and terms) to track in the series
target_entities = [
    'صدام', 'البعث', 'هدام', 'حزب', 'العراق','الحلقه الاخيره',
    'حسن', 'علي', 'محمد', 'بهاء',
    'بغداد', 'السجن', 'نهايه'
]

# 2. Function to search and extract entities from the text
def extract_mentions(text):
    found_entities = []
    # Check if text is a valid string
    if isinstance(text, str):
        words = text.split()
        for entity in target_entities:
            # If the entity exists in the comment, add it to our list
            if entity in words:
                found_entities.append(entity)
    return found_entities

# 3. Apply the function to create the new 'mentions' column
df_all['mentions'] = df_all['text'].apply(extract_mentions)

# 4. Print the first 5 rows to verify success
print("The 'mentions' column created successfully! Here is a sample:")
print(df_all[['text', 'mentions']].head(5))
```
